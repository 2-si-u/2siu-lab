## Series 함수
    - ndim : 차원 반환
    - len : 데이터 길이
    - size : 개수 반환
    - shape : 튜플의 형태로 shape반환
    - unique : 유일한 값만 udarray로 반환
    - count : NaN을 제외한 개수를 반환
    - mean : NaN를 제외한 평균
    - value_counts : NaN를 제외하고 각 값들의 빈도를 반환

## 시리즈(Series)의 연산
- 시리즈는 시리즈끼리 연산이 가능하다. 단 indax를 기주
- 이 특징은 DataFrame에서도 동일하게 적용된다

```python
hobbies = {
    "승현": "칵테일방, 찜질방, 노래방",
    "예나": "요리, 노래방, 영화",
    "진성": "호프, 노래방, 호프집",
    "가나디": "그릭요거트, 그림그리기, 영화",
    "지혁": "음주, 흡연, 도박",
    "시우": "볼링, 클라이밍, 보석십자수"
}

s4 = pd.Series(hobbies)
```

```python
# 악의 축 취미를 가진 사용자를 mask로 제거하기 (호프, 도박, 흡연)
s4
```

```python
evil_list = ["호프", "도박", "흡연"]
evil_mask = [sum(evil in hobby for evil in evil_list) == 0 for hobby in s4]

evil_member = s4[evil_mask]
print(evil_member)
```

```python
for hobby in s4:
    for evil in evil_list:
        print(evil in hobby)
    print()
```

```python
# contains
# s4: 각행을 접근해서 글자들이 포함되어있으면 True, 아니면 False
s4[s4.str.contains("호프|도박|흡연")]
s4[s4.str.contains("호프|도박|흡연") == 0]
```

```python
salary_list = [
    "지혁", "희수", "승현", "우영", "희수", "시우", "슬기", "상아", "상아", "희수", "우영", "진휘", 
    "지혁", "준승", "예나", "진휘", "지호", "지혁", "희수", "승현", "시우", "준승", "우영", "예나"
]

salary_income = [
    200, 300, 400, 400, 350, 500, 300, 600, 600, 500, 600, 800,
    300, 300, 500, 500, 300, 400, 300, 100, 300, 200, 700, 500,
]
```

```python
s1 = pd.Series(data=salary_income, index=salary_list)
s1
```

```python
# count(): 개수
s1.groupby(s1.index).count()
```

```python
# 더하는 집계함수
s1.groupby(s1.index).sum()
```

```python
# mean:평균
s1.groupby(s1.index).mean()
```

```python
s1.groupby(s1>300).count()
```

```python
s1.groupby(s1>300).mean()
```

```python
s1.groupby(s1 >= 600).apply(list)
```

```python
# to~: 형변환
# is~: bool
# get~: getter 값 반환
# set~: setter값 반환

print(s1.to_dict())
print(s1.to_list())
```
