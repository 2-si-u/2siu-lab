### shape
    - numpy에서는 해당 array의 크기를 알 수 있다. shape를 확인함으로서, 몇 개의 데이터가 있는지 몇 차원으로 존재하는지 확인할 수 있다.
    - arr.shape의 결과는 튜플로 (행, 열)로 행만 존재하면 1차원, 행렬이 존재하면 2차원으로 차수를 알 수 있다. 또한 행*렬을 곱하면 크기도 확인할 수 있다.

```python
print(arr3)
print(arr3.shape)


#(10,)
#1차원, 행의 10개 데이터
```

## seed()
- 난수를 예측가능하도록 만들며, 실행할 떄마다 동일한 난수를 발생시켜
   코드의 재현성, 특정 알고리즘의 결과를 재현하게 만든다.
   또한 해당 알고리즘의 결과의 오류를 추적하기 위해 사용한다.

```python
np.random.seed(4885)

np.random.randint(10, size=(2,5))
```

## 브로드캐스팅
- 다른 크기의 배열들 간에도 자동으로 연산해주는 Numpy의 강력한 기능

## 브로드캐스팅의 특징
    - 기본적으로는 shape와 같은 두 ndarray에 대한 연산은 각 원소별로 진행한다
    - 다른 shape를 갖는 array 간의 연산의 경우 브로드 캐스팅(shape를 맞춤) 후 진행한다

<img width=500 src='https://www.astroml.org/_images/fig_broadcast_visual_1.png' />

```python
arr1 = np.arange(12).reshape(4,3)
arr1

print(arr1.shape)
```

```python
arr2 = np.arange(3)

print(arr2)
print(arr2.shape)
```

```python
"""
  차원에 1이 존재하는 경우 확장이 가능하다.
    6 x 1 x 5
        7 x 3

    6 x 7 x 5

"""

None
```

```python
# (4,3)
arr2 = np.arange(12).reshape
```

<img src="https://numpy.org/doc/stable/_images/broadcasting_3.png" />

```python
# (2,3),(3,2)

arr3 = np.arange(6).reshape(2,3)

arr4 = np.arange(100, 106).reshape(3,2)


print(arr3)
print(arr4)
```

#### 열의 개수가 같아야, 계산이 가능하다

```python
arr3 + arr4
```

#### 아니면 계산이 가능하도록 차원을 바꾸주면 된다 

```python
arr3.T + arr4
```

## 차원축(axis)
- 1차원 array라면 각 차원에 대한 axis 값은 0
- 2차원 array라면 각 차원에 대한 axis 값은 0, 1
- 3차원 array라면 각 차원에 대한 axis 값은 0, 1, 2

```python
arr1 = np.arange(15)
arr1
```

```python
np.sum(arr1, axis=0)
```

```python
arr1 = np.arange(15).reshape(3,5)
arr1
```

```python
np.sum(arr1, axis=0)
```

```python
np.sum(arr1, axis=1)
```

```python
arr2 = np.arange(24).reshape(2,3,4)  # 3차원 데이터
arr2
```

```python
np.sum(arr2, axis=0)
```

## ※argmax() 중요!!!!!!!!!!!!!!!!!!!!!!!!!!!!



- 최대값의 인덱스

```python
arr3 = np.arange(5, 10)
print(arr3)
print(arr3[arr3.argmax()])
```

```python
np.random.seed(918)

arr4 = np.random.randint(10, size=(1,6)).flatten()

print(arr4.argmax())  # 가장 큰 값의 인덱스
print(arr4.argmin())  # 가장 작은 값의 인덱스
print(arr4.max())  # 가장 큰 값
print(arr4.min)  # 가장 작은 값

print(np.argmax(arr4))
```

```python
arr5 = np.random.randint(10, size=(3,6))
arr5
```

```python
print(arr5.argmax(axis=0))  # 접근 방법이 어디를 접근하는지에 따라서 달라질 수 있다 
```

```python
print(arr5.argmax(axis=1))
```

```python
print(arr5.min(axis=0))
```

## 다중조건식
- &: and 조건식
- | :or 조건식

```python
# 홀수이면서 50보다 작은 값
# and:

np.random.seed(2026)
arr5 = np.random.randint(100, size=15)
arr5
```

```python
is_bound= arr5 % 2 == 1 & (arr5 < 50)
arr5[is_bound]
```

```python
arr5[is_bound].astype(str)
```
