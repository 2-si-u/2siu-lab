```python
# 만약 NaN이라면: 알수없음
# 19세 이상: 성인
# 19세 미만: 아이
# id ~ elif ~ else

# 경우의 수가 세개니까 삼항은 사용할 수 없고 if-else가 더 유용하다. 단 선택해야할 것은 람다를 쓰게되면 삼항은 사용할 수 있는대 이프문은 사용하시 어렵다
# 이런경우 함수로 뺴놓고 작업한다

def get_adult(age):
    if np.isnan(age):
        return "알 수 없음"

    return "성인" if age >= 19 else "아이"



get_adult(np.nan)
```

```python
# apply(): 앞의 시리즈의 값을 순차적으로 받아서 결과를 바꾸고 새로운 결과를 반환하는 함수

titanic_df["IsAdult"] = titanic_df["Age"].apply(get_adult)
titanic_df
```

```python
# Ripoff
# Fare > 300: 바가지 요금
# Fare <= 300: 정상요금 


titanic_df["Ripoff"] = titanic_df["Fare"].apply(lambda fare: "바가지" if fare > 300 else "정상")
```

```python
titanic_df[titanic_df["Fare"] > 300]
```
