# 넘파이(Numpy)

    - 넘파이로만으로도 많이 사용되지만, 판다스(pandas), 머신러닝 들 수많은 분야에서

[넘파이 공식 문서](https://numpy.org/)

### 1) 수학 수치연산(numerical python)

### 2) 기본 타입

    - 제공함수 : 선형대수, 통계관련 여러 수치 함수를 제공한다
    - 그 외 수 많은 모듈에서 numpy 함수명과 기능을 그대로 사용 가능하다

파이썬과 다점 ->  타입을 강제로 승격 시킨다는 것 / 값을 가져왔을때 타입이 다르다는것 

```python
import numpy as np 
```

```python
arr1 = np.ndarray([1,2,3])

print(arr1)
print(type(arr1))
```

```python
arr1 = np.array([1,2,3])

print(arr1)
print(type(arr1))  # 콤마가 없지만 값은 나누어져 있음
```

```python
arr2 = np.array([1,2,3])

arr2
```

```python
# 하나의 타입으로만 정의할 수 있다.
arr2 = np.array([      1,2,3.0])

print(arr2)
print([1,2,3.0])
```

```python
# 넘파이의 타입 강제 승격: 전부 문자열로 출력-> 이들을 하나로 정의할 수 있는 것이 문자열 밖에 없음
arr3 = np.array([1,2,False,"홍길동", 12.5])

print(arr3)
```

```python
arr3 = np.array([i + 1 for i in range(10)])

print(arr3)
print("---------")
arr3
```

## ndarry의 인덱싱

```python
print(arr3[1])
print(type(arr3[1]))
print(int(arr3[1]))  # 강제형변환

print("----------")
print(2)
print(type(2))
```

## ndarray의 슬라이싱

```python
print(arr3) 
print(arr3[2:5])
print(arr3[::2]) 
```

### shape
    - numpy에서는 해당 array의 크기를 알 수 있다. shape를 확인함으로서, 몇 개의 데이터가 있는지 몇 차원으로 존재하는지 확인할 수 있다.
    - arr.shape의 결과는 튜플로 (행, 열)로 행만 존재하면 1차원, 행렬이 존재하면 2차원으로 차수를 알 수 있다. 또한 행*렬을 곱하면 크기도 확인할 수 있다.

```python
print(arr3)
print(arr3.shape)


#(10,)
#1차원, 행의 10개 데이터
```

```python
arr4 = np.array([2])
arr4
```

```python
# 2차원 스칼라 

print(arr4)
print(arr4.shape)
```

```python
arr5 = np.array([1,2,3],[4,5,6])

print(arr5)
print(arr5.shape)
#2행 3열 (2차원)
```

```python
arr6 = np.array([[1,2,3],[4,5,6]])


print(arr6)
print(arr6.shape)
```

```python
print(arr6.shape)

#텐서
#2행3열
```
